```python
df["nationality"]=df["nationality"].fillna("NA")
df["url"]=df["url"].fillna("NA")
```

```python
#deleting  unnecessary columns
df.drop(columns=['number','code'],inplace=True)
```

```python
# Convert to datetime
df['dob'] = pd.to_datetime(df['dob'], errors='coerce')

# Calculate median date
median_dob = df['dob'].median()

# Fill missing values
df['dob'] = df['dob'].fillna(median_dob)
```

# Fixing the Text encoding

```python
#checking the wrong encoding in names
df[df['forename'].str.contains('Ã', na=False)]
df[df['surname'].str.contains('Ã', na=False)]
```

```python
#fixing encoding
def fix_encoding(text):
    try:
        return text.encode('latin1').decode('utf-8')
    except:
        return text

for col in ['forename', 'surname']:
    df[col] = df[col].apply(fix_encoding)
```
